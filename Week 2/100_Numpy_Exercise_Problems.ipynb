#### 38. Consider a generator function that generates 10 integers and use it to build an array (★☆☆)

````python
def generate():
    values = []
    for x in range (10):
        values.append(x)
    return values

np.array(generate())
````

````python
# using yeild 
def generate2():
    yield from range(10)
        
for x in generate2():
    print(x) # we created an iterator 
    
arr1 =np.fromiter(generate2(), dtype=int) # we can create an array using this generator 
arr1
````

#### 47. Given two arrays, X and Y, construct the Cauchy matrix C (Cij =1/(xi - yj))

````python
# Normal way without numpy 
x= [1,2,3]
y= [4,5,6]
arr= np.zeros((len(x) ,len(y)))
arr
````

````python
for i in range(len(x)):
    for j in range(len(y)):
        value= 1 / (x[i] - y[j])
        arr[i,j]=value
        
arr
````

````python
np.linalg.det(arr) # deteminant calculation 
````

````python
#using numpy array 
faster_arr=1/ np.subtract.outer(x,y)
print(np.linalg.det(faster_arr))
````
